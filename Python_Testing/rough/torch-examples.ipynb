# Cool Pytorch Staffs

```python
# Imports
import torch
```

```python
randint=torch.randint(-100,100,[6])
# .randint() expects the size argument as a tuple=(6,) or list [6] rather a fat old int by itselfonly
randint
```

```python
# Utilizing Shapes of a Random Float Matrix
tensor=torch.rand(3,2)
tensor
```

```python
# zeros matrix, but all the zeros are floating point numbers
zeros=torch.zeros(2,3)
zeros
```

```python
# ones matrix are same as zeros matrix but 1.0 instead
ones=torch.ones(3,4)
ones
```

```python
# These are random values spanning a huge range, ipto avogadro exponents
empty=torch.empty(2,3)
empty
```

```python
# Arange follows the pythn range() function's upper limit member exclusion
arange=torch.arange(5)
# tensot([0to4])
arange
```

```python
# torch.linspace(), name is weird. Here range includes upper limit but step means (steps-1) division between the lower and upper limits
linspace=torch.linspace(3,10,steps=5)
# Note here, kwargs steps(as opposed to kwargs step) doesnt mean increaments
linspace
```

```python
# Logspace might be good for decibels, all the generators takes kwargs start=, end= but also first two args act as start and end
logspace=torch.logspace(start=-20,end=20,steps=5)
logspace
```

```python
# eye gives unit square matrix with arg as shape=arg*arg
eye=torch.eye(5)
eye
```

```python
# empty_like() makes a tensor similar to the args tensor in shape but populated with zeros
a=torch.empty((2,3),dtype=torch.int64)
empty_like=torch.empty_like(a)
empty_like
# It ahould have given me zeros but it did not
```
