```python
import music21

# 1. Create a Measure object
m = music21.stream.Measure(number=1)
m.timeSignature = music21.meter.TimeSignature('3/4')

# 2. Add your notes / tuplet rhythms into the measure
n1 = music21.note.Note('C4', quarterLength=1.0)
n2 = music21.note.Note('D4', quarterLength=1.0)
n3 = music21.note.Note('E4', quarterLength=1.0)
m.append([n1, n2, n3])

# 3. Add the measure to your main Stream or Part
my_stream = music21.stream.Stream()
my_stream.append(m)

# 4. Export as MIDI
my_stream.write('midi', fp='output.mid')
print("MIDI exported successfully!")
```
