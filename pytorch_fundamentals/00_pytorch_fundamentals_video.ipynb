## Random Tensors

### Why Random Tensors?

Random tensors are important  because the way many neural networks learn is that they start with tensors full of random numbers and then adjust those random numbers to better represent the data.

`Start with random numbers -> look at data -> update random numbers -> look at data -> update random numbers`

Torch random tensors: https://pytorch.org/docs/stable/generated/torch.rand.html

```python
10*10*10
```

```python
# Create a random tensor of size (3,4)
random_tensor=torch.rand(10,10,10)
random_tensor
# Probably using random module in the background
```

### Tensor Datatypes

**Note:** Tensor datatypes or so called dtypes is one of the 3 big errors we may run into with PyTorch & deep learning:
1. Tensors not right dtypes `Low Level mmismatch Between float32 and float16|float64`
2. Tensors not right shape `Linear Algebra Errors`: `MATRIX multiplication not lining up`
3. Tensors not on right device









```python
# Float 32 tensor
float_32_tensor=torch.tensor([3.0,6.0,9.0],
                             dtype=torch.float32, #dtypes are the types of numbers we can feed into the tensor. The difference between float16 and float64 are related to the precision in computing and the extent of detail in which the quantity is expressed
                             device='cpu', # We can use cpu or gpu(cuda). We will later touch on device diagnostic tests.
                             requires_grad=False # If we want pytorch to track the gradients, which we will cover later, of a tensor when we are dealing with certain numerical calculations
                             )
  # All the very important kwargs for torch.tensor()
  # Instead of dtype=None can also specify dtype=torch.float16(floats comprising upto 16bit for the decimal points inside ram, why ram is so valuable these days); remember the dtypes are also torch objects
float_32_tensor
```

```python
float_32_tensor.dtype
# The default datatype of tensors even if kw dtype=None; it will still have dtype=float32
```

```python
float_16_tensor=float_32_tensor.type(torch.float16)
float_16_tensor
# We can use half or double in place of float of torch.float which indicates float32
float_16_tensor=float_32_tensor.type(torch.half)
float_16_tensor


```

```python
float_32_tensor*float_16_tensor
```

```python
float_16_tensor*float_32_tensor
```

```python
data=[random.randrange(1,9) for _ in range(3)]
int_32_tensor=torch.tensor(data,dtype=torch.int32)

int_32_tensor
```

```python
int_32_tensor*float_32_tensor
# All works, even torch.int64 works,we can also use torch.long which called torch.long
# But matrix of different dimension throws error for sure, a linear algebra situation
```

### Manipulating Tensors `tensor_operations`

Tensor operations include:
* `Addition`
* `Subtraction`
* `Multiplication` (element-wise)
* `Division`
* `Matrix` Multiplication

To find patterns in the number of dataset a Neural Network will combine the above functions in someway shapes or form.










```python
# @title Addition
# Create a tensor and add 10 to it
tensor=torch.tensor([random.randrange(-9,9) for _ in range(3)],dtype=torch.int32)
tensor,tensor+10
```

```python
# @title Subtraction
# Create a tensor and subtract 10 from it
tensor=torch.tensor([random.randrange(-9,9) for _ in range(3)],dtype=torch.int32)
tensor,tensor-10
```

```python
# @title Multiplication
# Create a tensor and multiply with 10
tensor=torch.tensor([random.randrange(-9,9) for _ in range(3)],dtype=torch.int32)
tensor,tensor*10
```

```python
# @title Division
# Create a tensor and add 10 to it
tensor=torch.tensor([random.randrange(0,9) for _ in range(3)],dtype=torch.int32)
tensor,tensor/10
```

```python
# Try out PyTorch in-built functions
torch.add(tensor,10),torch.sub(tensor,10),torch.mul(tensor,10),torch.div(tensor,10)
```

```python
# @title Addition
# Create a tensor and add 10 to it
tensor=torch.tensor([random.randrange(-9,9) for _ in range(3)],dtype=torch.int32)
tensor,tensor+10
```
